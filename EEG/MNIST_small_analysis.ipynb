```python
import re
from matplotlib import colormaps
import matplotlib.pyplot as plt
import mne

# 1. Identify Left/Right pairs and Midline channels dynamically
ch_names = epochs.ch_names

# Standard pairings: Left has odd numbers, Right has even numbers
left_channels = sorted(
    [ch for ch in ch_names if re.search(r"[13579]$", ch)],
    key=lambda x: re.sub(r"\d+$", "", x),
)

pairs = []
used = set()
for left in left_channels:
    prefix = re.sub(r"\d+$", "", left)
    num = int(re.search(r"\d+$", left).group())
    right = f"{prefix}{num + 1}"
    if right in ch_names:
        pairs.append((left, right))
        used.update([left, right])

# Optional: Pair midline channels (e.g., Fz, Cz, Pz, Oz) if present
midline_channels = [ch for ch in ch_names if ch.lower().endswith("z")]
midline_pairs = [
    (midline_channels[i], midline_channels[i + 1])
    for i in range(0, len(midline_channels) - 1, 2)
]
if len(midline_channels) % 2 != 0:
    midline_pairs.append((midline_channels[-1], None))

all_pairs = pairs + midline_pairs
n_rows = len(all_pairs)

# 2. Build the subplot layout: n_rows x 2 columns
fig, axes = plt.subplots(
    nrows=n_rows,
    ncols=2,
    figsize=(14, 3.2 * n_rows),
    sharex=True,
    sharey=True,
)

# Ensure 2D array if there is only 1 pair
if n_rows == 1:
    axes = axes.reshape(1, 2)

palette = colormaps.get_cmap("tab20").colors

# 3. Plot each channel into its corresponding axis
for row_idx, (left_ch, right_ch) in enumerate(all_pairs):
    ax_left = axes[row_idx, 0]
    ax_right = axes[row_idx, 1]

    # Show legend only on the first plot (top-right) to keep axes clean
    legend_pos = "upper right" if row_idx == 0 else False

    # Left Channel
    if left_ch:
        mne.viz.plot_compare_evokeds(
            evokeds,
            picks=[left_ch],
            combine="mean",
            title=f"{left_ch} ERP",
            colors=palette,
            axes=ax_left,
            legend=False,
            show=False,
        )

    # Right Channel
    if right_ch:
        mne.viz.plot_compare_evokeds(
            evokeds,
            picks=[right_ch],
            combine="mean",
            title=f"{right_ch} ERP",
            colors=palette,
            axes=ax_right,
            legend=legend_pos,
            show=False,
        )
    else:
        ax_right.axis("off")  # Blank if an odd number of midline electrodes

    # Add column headers on the very top row
    if row_idx == 0:
        ax_left.set_title(
            f"Left Hemisphere\n{ax_left.get_title()}",
            fontsize=12,
            fontweight="bold",
        )
        ax_right.set_title(
            f"Right Hemisphere\n{ax_right.get_title()}",
            fontsize=12,
            fontweight="bold",
        )

plt.tight_layout()
plt.show()
```

```python
# 1. Montage and Evokeds Setup
montage = mne.channels.make_standard_montage("colin27_1020")
digits = [str(d) for d in range(-1, 10)]  # ['-1', '0', ..., '9']

for d in digits:
    evokeds[d].set_montage(montage)

# 2. Define Animation Timing
times = np.linspace(0.0, 0.5, 50)  # 50 frames from 0ms to 500ms
fps = 15

# Determine global min/max across all digits and times for a consistent colorbar
all_data = np.stack(
    [evokeds[d].get_data(tmin=0.0, tmax=0.5) for d in digits]
)  # (n_digits, n_channels, n_times)
# Scale to microvolts (µV)
all_data_uv = all_data * 1e6
vlim = (np.percentile(all_data_uv, 2), np.percentile(all_data_uv, 98))
# Alternatively symmetric:
# max_abs = np.percentile(np.abs(all_data_uv), 98)
# vlim = (-max_abs, max_abs)

# 3. Create Subplot Layout (e.g., 3 rows x 4 columns = 12 axes for 11 digits)
fig, axes = plt.subplots(3, 4, figsize=(16, 12))
axes_flat = axes.flatten()

# Turn off the 12th unused axis
axes_flat[-1].axis("off")

# Pre-extract channel positions and info from one evoked
info = evokeds[digits[0]].info

# Super title showing the evolving time
time_title = fig.suptitle(
    f"Time: {times[0]*1000:.1f} ms", fontsize=18, fontweight="bold"
)


# 4. Animation Update Function
def update(frame_idx):
    current_time = times[frame_idx]
    time_title.set_text(f"Visual ERP Latency: {current_time*1000:.0f} ms")

    for idx, d in enumerate(digits):
        ax = axes_flat[idx]
        ax.clear()

        # Extract channel values at the exact time frame
        evoked_t = evokeds[d]
        time_point_idx = np.argmin(np.abs(evoked_t.times - current_time))
        data_t = evoked_t.data[:, time_point_idx] * 1e6  # in µV

        # Plot 2D scalp topomap into current subplot axis
        mne.viz.plot_topomap(
            data_t,
            info,
            axes=ax,
            show=False,
            vlim=vlim,
            cmap="RdBu_r",
            contours=4,
            sensors=True,
        )

        label = "Baseline / Noise (-1)" if d == "-1" else f"Digit {d}"
        ax.set_title(label, fontsize=12, fontweight="bold")

    return axes_flat


# 5. Build and Save Animation
anim = animation.FuncAnimation(
    fig, update, frames=len(times), interval=1000 / fps, blit=False
)

# Optional: Add a shared scalar colorbar to the figure
norm = plt.Normalize(vmin=vlim[0], vmax=vlim[1])
sm = plt.cm.ScalarMappable(norm=norm, cmap="RdBu_r")
cbar_ax = fig.add_axes([0.78, 0.12, 0.02, 0.20])  # [left, bottom, width, height]
cbar = fig.colorbar(sm, cax=cbar_ax)
cbar.set_label("Voltage (µV)", fontsize=11)

plt.tight_layout(rect=[0, 0, 0.95, 0.95])

# Save as GIF or MP4
anim.save("all_digits_topomap_animation.gif", writer="pillow", fps=fps, dpi=150)
# Or for higher quality/faster rendering:
# anim.save("all_digits_topomap_animation.mp4", writer="ffmpeg", fps=fps, dpi=150)

plt.show()
```

```python
def delay_embed(signal, tau=6, m=3):
    """Reconstruct phase space using delay-coordinate embedding (Takens' Theorem)."""
    N = len(signal) - (m - 1) * tau
    return np.array([signal[i : i + N] for i in range(0, m * tau, tau)]).T


# 1. Define digits and target electrode
digits = [str(d) for d in range(-1, 10)]  # ['-1', '0', '1', ..., '9']
target_channel = "O1"  # Primary visual cortex electrode
tau = 6  # Embedding delay (samples)
m = 3  # Embedding dimension

# 2. Set up a 3-row x 4-column 3D subplot grid (12 panels for 11 digits)
fig = plt.figure(figsize=(18, 13))
cmap = plt.cm.tab10

# Find global coordinate limits across all digits so all attractors share the same scale
all_trajectories = {}
for digit in digits:
    sig = evokeds[digit].get_data(picks=target_channel)[0] * 1e6  # Convert to µV
    all_trajectories[digit] = delay_embed(sig, tau=tau, m=m)

# Compute unified axes limits (with 5% padding)
all_coords = np.vstack(list(all_trajectories.values()))
coord_min, coord_max = all_coords.min(), all_coords.max()
padding = (coord_max - coord_min) * 0.05
lims = (coord_min - padding, coord_max + padding)

# 3. Loop over digits and plot each 3D attractor
for idx, digit in enumerate(digits):
    ax = fig.add_subplot(3, 4, idx + 1, projection="3d")
    traj = all_trajectories[digit]

    # Assign distinct colors per digit
    color = "gray" if digit == "-1" else cmap(int(digit) % 10)

    # Plot attractor trajectory
    ax.plot(traj[:, 0], traj[:, 1], traj[:, 2], lw=1.0, color=color, alpha=0.85)

    # Mark the trajectory origin (t = 0ms)
    ax.scatter(
        traj[0, 0],
        traj[0, 1],
        traj[0, 2],
        color="black",
        s=25,
        label="Onset (t=0)",
    )

    # Formatting and labels
    label = "Baseline / Noise (-1)" if digit == "-1" else f"Digit {digit}"
    ax.set_title(
        f"{label} ({target_channel})", fontsize=11, fontweight="bold", pad=2
    )

    # Keep scales consistent for direct comparison
    ax.set_xlim(lims)
    ax.set_ylim(lims)
    ax.set_zlim(lims)

    ax.set_xlabel(r"$x(t)$", fontsize=8, labelpad=-2)
    ax.set_ylabel(r"$x(t+\tau)$", fontsize=8, labelpad=-2)
    ax.set_zlabel(r"$x(t+2\tau)$", fontsize=8, labelpad=-2)
    ax.tick_params(labelsize=7)
    ax.view_init(elev=25, azim=45)  # Set consistent camera viewing angle

# Hide the unused 12th panel
ax_unused = fig.add_subplot(3, 4, 12)
ax_unused.axis("off")

plt.suptitle(
    f"3D Phase Space Attractor Convergence across Digits (Electrode {target_channel}, $\\tau={tau}$, $m={m}$)",
    fontsize=15,
    fontweight="bold",
    y=0.98,
)
plt.tight_layout()
plt.show()
```

```python
import math
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np


def delay_embed(signal, tau=6, m=3):
    """Reconstruct phase space using delay-coordinate embedding."""
    N = len(signal) - (m - 1) * tau
    return np.array([signal[i : i + N] for i in range(0, m * tau, tau)]).T


# 1. Setup parameters
digits = [str(d) for d in range(-1, 10)]  # Starts with '-1', then '0' to '9'
channels = evokeds[digits[0]].ch_names
tau = 6
m = 3

cmap = plt.get_cmap("tab20")
digit_colors = {d: cmap(i / len(digits)) for i, d in enumerate(digits)}

# Determine subplot grid shape (rows x cols) based on channel count
n_chans = len(channels)
ncols = 4
nrows = math.ceil(n_chans / ncols)

fig = plt.figure(figsize=(5 * ncols, 4.5 * nrows))

# 2. Loop over each electrode
for ch_idx, ch in enumerate(channels):
    ax = fig.add_subplot(nrows, ncols, ch_idx + 1, projection="3d")

    # Plot each digit's trajectory on the same 3D axis
    for d in digits[1:]:
        sig = evokeds[d].get_data(picks=ch)[0] * 1e6  # µV
        traj = delay_embed(sig, tau=tau, m=m)

        label_name = f"Noise ({d})" if d == "-1" else f"Digit {d}"
        ax.plot(
            traj[:, 0],
            traj[:, 1],
            traj[:, 2],
            label=label_name,
            color=digit_colors[d],
            lw=1.0,
            alpha=0.75,
        )

        # Mark trajectory onset (t=0)
        ax.scatter(
            traj[0, 0], traj[0, 1], traj[0, 2], color=digit_colors[d], s=15
        )

    ax.set_title(f"Electrode: {ch}", fontsize=11, fontweight="bold")
    ax.set_xlabel(r"$x(t)$", fontsize=8, labelpad=-2)
    ax.set_ylabel(r"$x(t+\tau)$", fontsize=8, labelpad=-2)
    ax.set_zlabel(r"$x(t+2\tau)$", fontsize=8, labelpad=-2)
    ax.tick_params(labelsize=7)
    ax.view_init(elev=25, azim=45)

    # Place legend only on the first plot to avoid clutter
    if ch_idx == 0:
        ax.legend(
            bbox_to_anchor=(0.0, 1.15),
            loc="upper left",
            fontsize=7,
            ncol=3,
            frameon=True,
        )

plt.suptitle(
    f"Per-Electrode 3D Attractors across Digits (-1 to 9) [$\\tau={tau}$, $m={m}$]",
    fontsize=16,
    fontweight="bold",
    y=0.99,
)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import numpy as np


def delay_embed(signal, tau=6, m=3):
    """Reconstruct phase space using delay-coordinate embedding."""
    N = len(signal) - (m - 1) * tau
    return np.array([signal[i : i + N] for i in range(0, m * tau, tau)]).T


# 1. Setup parameters
digits = [str(d) for d in range(-1, 10)]  # Starts with '-1', then '0' to '9'
tau = 6
m = 3

cmap = plt.get_cmap("tab20")
digit_colors = {d: cmap(i / len(digits)) for i, d in enumerate(digits)}

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

# 2. Overlay all digits averaged across all electrodes
for d in digits[1:]:
    # Mean across all channels (shape: [times])
    sig_all_chans = evokeds[d].get_data().mean(axis=0) * 1e6  # in µV
    traj = delay_embed(sig_all_chans, tau=tau, m=m)

    label_name = f"Noise ({d})" if d == "-1" else f"Digit {d}"

    # Trajectory curve
    ax.plot(
        traj[:, 0],
        traj[:, 1],
        traj[:, 2],
        label=label_name,
        color=digit_colors[d],
        lw=1.6,
        alpha=0.85,
    )

    # Mark onset (t=0)
    ax.scatter(
        traj[0, 0],
        traj[0, 1],
        traj[0, 2],
        color=digit_colors[d],
        s=40,
        edgecolors="black",
        linewidths=0.5,
    )

# 3. Styling & Labels
ax.set_title(
    f"Grand Average 3D Attractor Convergence across All Electrodes\n(Digits -1 to 9, $\\tau={tau}$, $m={m}$)",
    fontsize=13,
    fontweight="bold",
    pad=15,
)
ax.set_xlabel(r"$x(t)$ [$\mu$V]", fontsize=10, labelpad=8)
ax.set_ylabel(r"$x(t+\tau)$ [$\mu$V]", fontsize=10, labelpad=8)
ax.set_zlabel(r"$x(t+2\tau)$ [$\mu$V]", fontsize=10, labelpad=8)

ax.view_init(elev=28, azim=50)
ax.legend(
    loc="center left",
    bbox_to_anchor=(1.05, 0.5),
    fontsize=9,
    title="Digit Class",
    frameon=True,
)

plt.tight_layout()
plt.show()
```
